## 4. Careful words: what I can and can't claim

### What I can claim

I can identify and describe **observed patterns and associations** between content characteristics and performance metrics such as impressions, clicks, pageviews, and sessions. I can use these findings to provide **directional evidence and decision support** for deciding which content may deserve further review or optimization.

### What I cannot claim

Any recommendations should be treated as decision support and validated before making significant content or resource-allocation decisions.
